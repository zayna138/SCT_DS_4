# SCT_DS_4 - Traffic Accident Data Analysis

This project analyzes traffic accident data to identify patterns and trends using Python.

## Objective
Analyze accident data and visualize important insights related to road accidents.

## Libraries Used
- Pandas
- Matplotlib

## Analysis Performed
- Accident distribution analysis
- Severity analysis
- Weather condition analysis
- Time-based accident trends

## Output
- Bar charts
- Histograms
- Data insights

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("US_Accidents_March23.csv")

print(df.head())

print(df.info())

print(df.isnull().sum())

# Severity Count
severity_count = df['Severity'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(severity_count.index,
        severity_count.values,
        color=['skyblue', 'orange', 'mediumseagreen', 'tomato'])

plt.title("Accident Severity Distribution")
plt.xlabel("Severity")
plt.ylabel("Count")

plt.show()

# Weather Conditions
weather = df['Weather_Condition'].value_counts().head(10)

plt.figure(figsize=(10,5))
plt.bar(weather.index,
        weather.values,
        color='slateblue')

plt.title("Top 10 Weather Conditions During Accidents")
plt.xlabel("Weather Condition")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.show()

# Temperature Distribution
plt.figure(figsize=(8,5))
plt.hist(df['Temperature(F)'].dropna(),
         bins=20,
         color='brown',
         edgecolor='black')

plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")

plt.show()
```
